```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import math

# 1. Recriar os pontos de entrega
dados = {
    "Ponto": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Local": [
        "Restaurante", "Praça Central", "Bairro Norte",
        "Bairro Sul", "Bairro Leste", "Bairro Oeste",
        "Bairro Novo", "Centro Comercial"
    ],
    "Coordenada X": [0, 2, 4, 1, 5, 0, 3, 6],
    "Coordenada Y": [0, 1, 2, 3, 4, 5, 5, 1]
}

pontos = pd.DataFrame(dados)

# 2. Recriar as ruas
ruas = {
    "Origem": ["A", "A", "B", "B", "B", "C", "D", "D", "E", "F", "G"],
    "Destino": ["B", "D", "C", "D", "H", "E", "F", "G", "G", "G", "H"],
    "Distancia": [2.2, 3.2, 2.2, 2.5, 4.1, 2.5, 2.2, 3.0, 2.2, 3.2, 4.0]
}

df_ruas = pd.DataFrame(ruas)

# 3. Salvar os arquivos CSV novamente
pontos.to_csv("pontos_entrega.csv", index=False)
df_ruas.to_csv("ruas.csv", index=False)

# 4. Construir o grafo
G = nx.Graph()

for _, linha in df_ruas.iterrows():
    G.add_edge(
        linha["Origem"],
        linha["Destino"],
        weight=linha["Distancia"]
    )

# 5. Definir as posições dos pontos
posicoes = {
    linha["Ponto"]: (
        linha["Coordenada X"],
        linha["Coordenada Y"]
    )
    for _, linha in pontos.iterrows()
}

# 6. Exibir os dados
print("Pontos e ruas recriados com sucesso!")
print("Quantidade de pontos:", G.number_of_nodes())
print("Quantidade de conexões:", G.number_of_edges())

display(pontos)
display(df_ruas)
```

```python
# Estimar a distância em linha reta
def heuristica(origem, destino):
    x1, y1 = posicoes[origem]
    x2, y2 = posicoes[destino]

    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)


# Calcular a rota de menor custo
def encontrar_rota(origem, destino):
    rota = nx.astar_path(
        G,
        origem,
        destino,
        heuristic=lambda atual, objetivo: heuristica(
            atual, objetivo
        ),
        weight="weight"
    )

    distancia = nx.path_weight(
        G, rota, weight="weight"
    )

    return rota, distancia


# Testar três rotas
testes = [("A", "C"), ("A", "E"), ("A", "F")]

resultados = []

for origem, destino in testes:
    rota, distancia = encontrar_rota(origem, destino)

    print(f"Origem: {origem}")
    print(f"Destino: {destino}")
    print(f"Rota: {' -> '.join(rota)}")
    print(f"Custo total: {distancia:.2f}")
    print("-" * 30)

    resultados.append({
        "Origem": origem,
        "Destino": destino,
        "Caminho": " -> ".join(rota),
        "Custo_Total": round(distancia, 2)
    })

# Salvar os resultados
df_resultados = pd.DataFrame(resultados)
df_resultados.to_csv("resultados_rotas.csv", index=False)

print("Resultados salvos em resultados_rotas.csv")

display(df_resultados)
```

```python
import math
import networkx as nx
import pandas as pd

# Calcular a distância estimada entre dois pontos
def heuristica(origem, destino):
    x1, y1 = posicoes[origem]
    x2, y2 = posicoes[destino]

    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)


# Encontrar a melhor rota usando o algoritmo A*
def encontrar_rota(origem, destino):
    rota = nx.astar_path(
        G,
        origem,
        destino,
        heuristic=lambda atual, objetivo: heuristica(atual, objetivo),
        weight="weight"
    )

    distancia = nx.path_weight(G, rota, weight="weight")

    return rota, distancia


# Testar três rotas
testes = [
    ("A", "C"),
    ("A", "E"),
    ("A", "F")
]

resultados = []

for origem, destino in testes:
    rota, distancia = encontrar_rota(origem, destino)

    print("Origem:", origem)
    print("Destino:", destino)
    print("Rota:", " -> ".join(rota))
    print("Custo total:", round(distancia, 2))
    print("-" * 30)

    resultados.append({
        "Origem": origem,
        "Destino": destino,
        "Caminho": " -> ".join(rota),
        "Custo_Total": round(distancia, 2)
    })

# Criar e salvar a tabela
df_resultados = pd.DataFrame(resultados)

df_resultados.to_csv("resultados_rotas.csv", index=False)

print("Resultados salvos com sucesso!")

display(df_resultados)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Selecionar as coordenadas dos pontos
coordenadas = pontos[["Coordenada X", "Coordenada Y"]]

# Criar três agrupamentos
modelo_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Identificar o grupo de cada ponto
pontos["Grupo"] = modelo_kmeans.fit_predict(coordenadas)

# Mostrar os resultados
display(pontos)

# Salvar os agrupamentos em CSV
pontos.to_csv("agrupamentos_entregas.csv", index=False)

print("Agrupamentos salvos com sucesso!")
```

```python
plt.figure(figsize=(10, 6))

for grupo in sorted(pontos["Grupo"].unique()):
    dados_grupo = pontos[pontos["Grupo"] == grupo]

    plt.scatter(
        dados_grupo["Coordenada X"],
        dados_grupo["Coordenada Y"],
        s=120,
        label=f"Grupo {grupo}"
    )

    for _, linha in dados_grupo.iterrows():
        plt.annotate(
            linha["Ponto"],
            (linha["Coordenada X"], linha["Coordenada Y"]),
            xytext=(5, 5),
            textcoords="offset points"
        )

plt.title("Agrupamento dos Pontos de Entrega com K-Means")
plt.xlabel("Coordenada X")
plt.ylabel("Coordenada Y")
plt.legend()
plt.grid(True)

plt.savefig("agrupamento_kmeans.png", dpi=300, bbox_inches="tight")
plt.show()

print("Gráfico salvo como agrupamento_kmeans.png")
```
